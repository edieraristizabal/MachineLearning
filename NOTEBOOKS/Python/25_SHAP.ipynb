# 25. Explicaciones aditivas de Shapley (SHAP) 

Los valores de SHAP dicen cuánto cada variable contribuye a la predicción. No es suficiente solo tener predicciones ajustadas; es necesario también explicar por qué el modelo hace tales predicciones. 

Aunque la regresión logística suele considerarse interpretable debido a su naturaleza lineal, todavía tiene algunos puntos ciegos. Los coeficientes crudos pueden darte una idea de la dirección (impacto positivo o negativo), pero no cuentan toda la historia. Por ejemplo, las interacciones entre variables o los efectos no lineales pueden oscurecer la imagen.

Aquí es donde entra SHAP para ofrecerte una visión más clara. Te permite cuantificar y visualizar la contribución exacta de cada característica, incluso cuando las relaciones no son sencillas. Al integrar los valores de SHAP con la regresión logística, obtienes lo mejor de ambos mundos: la simplicidad de un modelo lineal y la profundidad de una herramienta de interpretabilidad de vanguardia.

En términos prácticos, los valores SHAP te ayudan a responder dos preguntas importantes:

**Interpretabilidad Local**: ¿Por qué el modelo hizo esta predicción específica para esta persona en particular? SHAP te muestra qué características (por ejemplo, una baja puntuación de crédito) contribuyeron más a la predicción y en qué medida.

**Interpretabilidad Global**: ¿Qué características importan en general para el modelo en todas las predicciones? Los diagramas de resumen de SHAP y otras visualizaciones te dan una vista panorámica de cuáles características son las más influyentes.

## Interpretabilidad Local vs. Global
Aclaremos la diferencia entre interpretabilidad local y global, ya que es clave para aprovechar al máximo SHAP.

**Interpretabilidad Local**: Se centra en predicciones individuales. ¿Por qué tu modelo predijo que la persona A tiene un 75% de probabilidad de incumplir el pago de su préstamo? Los valores SHAP desglosan la influencia de cada característica (por ejemplo, un ingreso bajo aumentó la probabilidad en un 20%, mientras que un historial laboral estable la redujo en un 10%).

**Interpretabilidad Global**: En lugar de enfocarse en una sola predicción, la interpretabilidad global te ayuda a comprender el comportamiento del modelo en todo tu conjunto de datos. Podrás ver qué características suelen ser más influyentes. Por ejemplo, podrías descubrir que la puntuación de crédito es constantemente la característica más determinante para predecir el impago de préstamos en todos tus clientes.

Esta doble capacidad de SHAP lo convierte en una herramienta muy poderosa cuando necesitas tanto perspectivas granulares (locales) como tendencias generales (globales) en las predicciones de tu modelo.









```python
# Step 2: Import necessary libraries
import shap
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import statsmodels.api as sm
```

```python
# Code snippet from SHAP github page
import xgboost
import shap

# train an XGBoost model
X, y = shap.datasets.boston()
model = xgboost.XGBRegressor().fit(X, y)

# explain the model's predictions using SHAP
# (same syntax works for LightGBM, CatBoost, scikit-learn, transformers, Spark, etc.)
explainer = shap.Explainer(model)
shap_values = explainer(X)

# visualize the first prediction's explanation
shap.plots.waterfall(shap_values[0])
```

```python
# Code snippet from SHAP github page
# visualize the first prediction's explanation with a force plot
shap.plots.force(shap_values[0])
```

```python
shap.summary_plot(shap_values, X)
```
