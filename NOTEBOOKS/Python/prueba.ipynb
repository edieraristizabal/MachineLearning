```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import weibull_min, lognorm
# Time axis
t = np.linspace(0.1, 40, 400)
# Weibull parameters
k = 1.8
lambda_w = 15
weibull_survival = weibull_min.sf(t, k, scale=lambda_w)
# Log-normal parameters
sigma = 0.7
mu = 2.5
lognormal_survival = lognorm.sf(t, s=sigma, scale=np.exp(mu))
plt.figure(figsize=(8,4))
plt.plot(t, weibull_survival, label="Weibull", linewidth=2)
plt.plot(t, lognormal_survival, label="Log-Normal", linewidth=2)
plt.title("Weibull vs Log-Normal Survival")
plt.xlabel("Time")
plt.ylabel("Probability Still Active")
plt.legend()
plt.tight_layout()
plt.show()
```
